```python
#Exercicio 3.8 da página 50 sobre organizar lista
lugares = ['lugar', 'sei lá', 'ali', 'terra']
print("Esse é a lista de lugares que eu gostaria de estar na ordem original: ")
print(lugares)

#Mudando a ordem da lista com sorted()
print("Essa é a lista com a ordem alterada:")
print(sorted(lugares))

#Mostrando que a original se mantém
print("A lista original ainda se mantém:")
print(lugares)

#Mostrando a lista em ordem alfabetica reversa
print("Ordem alfabética ao contrario:")
print(sorted(lugares, reverse=True))

#Mostrando que a lista original ainda se mantém
print("A lista original ainda se mantém:")
print(lugares)

#utilizando reverse() para mudar a ordem da lista
print("Mudando a ordem com reverse:")
lugares.reverse()
print(lugares)

#Utilizando reverse de novo e ela volta para o formato original
print("Voltando para a original com reverse:")
lugares.reverse()
print(lugares)

#Utilizando sort() para mudar a ordem da lista
print("Mudando a ordem novamente:")
lugares.sort()
print(lugares)

#Utilizando sort() para colocar a lista em ordem alfabetica ao contrário
print("Colocando em ordem alfabetica ao contrario:")
lugares.sort(reverse=True)
print(lugares)
```

```python
#using if Statements with Lists
#Checking for Special Items p.90
#Utiliza um if dentro do for para verificar se há ingredientes validos para a pizza
#Por isso recebeu uma mensagem dizendo que não tem pimenta verde
requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
for requested_topping in requested_toppings:
  if requested_topping == 'green peppers':
    print('Sorry, we are out of green peppers right now')
  else:
    print('Adding ' + requested_topping + '.')
print('\nFinished making your pizza!')
```

```python
#Checking That a List Is Not Empty, p.91
#Aqui utilizamos o if antes do for para verificar se a lista está vazia
#Quando o nome da lista é chamado no if, se tiver algum item é de valor True
#Quando está vazia é False, portanto pula para a proxima condição, nesse caso else
requested_toppings = []
if requested_toppings:
  for requested_topping in requested_toppings:
    print('Adding ' + requested_topping + '.')
  print('\nFinished making your pizza!')
else:
  print('Are you sure you want a plain pizza?')
```

```python
#Using Multiple Lists
#Para verificar se itens de uma lista estão em outra lista
available_toppigns = ['mushrooms', 'olives', 'green peppers', 'pepperoni', 'pineaple', 'extra cheese']
requested_toppings = ['mushrooms', 'french fries', 'extra cheese']
for requested_topping in requested_toppings:
  if requested_topping in available_toppigns:
    print('Adding ' + requested_topping + '.')
  else:
    print('Sorry, we dont have ' + requested_topping + '.')
print('\nFinished making your pizza!')
```
